```python
# Librerias

from pyspark.sql import functions as F
import os
```

## Datos sobre caracteristicas

```python
# Ruta de la carpeta
ruta = "/Volumes/workspace/default/data/caracteristicas/"

# Obtener los archivos CSV
archivos = [
    os.path.join(ruta, archivo)
    for archivo in os.listdir(ruta)
    if archivo.upper().endswith(".CSV")
]

# Lista para almacenar los DataFrames
lista_datos = []

# Leer cada archivo
for archivo in archivos:
    datos = (
        spark.read
        .option("header", True)
        .option("sep", ";")
        .option("inferSchema", False)
        .csv(archivo)
    )

    # Convertir todas las columnas a string
    datos = datos.select([
        F.col(c).cast("string").alias(c)
        for c in datos.columns
    ])

    lista_datos.append(datos)

# Unir todos los DataFrames
Caracteristicas = lista_datos[0]

for datos in lista_datos[1:]:
    Caracteristicas = Caracteristicas.unionByName(
        datos,
        allowMissingColumns=True
    )

# Guardar tabla de caracteristicas 
Caracteristicas.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("caracteristicas")
```
